```python
#Step 1. Sample Size

data = {
    'Salary': [40000, 45000, 50000, 55000, 60000, 65000, 52000, 58000, 70000, 62000],
    'Education': ['High School', 'High School', 'Bachelor', 'Bachelor', 'Master',
                        'Master', 'Master', 'Bachelor', 'Bachelor', 'Master'],
    'Experience': [1, 2, 3, 4, 5, 6, 7, 3, 4, 5]
}
df = pd.DataFrame(data)
```

```python
model = smf.ols('Salary ~ Experience + C(Education, Treatment(reference = "High School"))', data = df).fit()
```

```python
print(model.summary())
```
