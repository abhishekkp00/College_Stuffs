### You want to estimate the mean salary of software engineers in a country. You take 10 different random samples, each containing 50 engineers, and calculate the sample mean for each. Plot the distribution of these sample means. How does the Central Limit Theorem explain the shape of this sampling distribution, even if the underlying salary distribution is skewed?

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Simulate population salaries (skewed distribution)
np.random.seed(42)
population_salaries = np.random.exponential(scale=50000, size=10000)  # skewed data
```

```python
# Number of samples and sample size
num_samples = 10
sample_size = 50
```

```python
# Calculate sample means
sample_means = []
for _ in range(num_samples):
    sample = np.random.choice(population_salaries, size=sample_size, replace=False)
    sample_means.append(float(np.mean(sample)))
```

```python
# Plot the distribution of sample means
plt.hist(sample_means, bins=5, edgecolor='black')
plt.title("Distribution of Sample Means (n=50, 10 samples)")
plt.xlabel("Sample Mean Salary")
plt.ylabel("Frequency")
plt.show()
```

```python
# Print sample means
print("Sample Means:", sample_means)
```
