### 7.	You are comparing the average daily sales between two stores. Store A has a mean daily sales value of $1,000 with a standard deviation of $100 over 30 days, and Store B has a mean daily sales value of $950 with a standard deviation of $120 over 30 days. Conduct a two-sample t-test to determine if there is a significant difference between the average sales of the two stores at the 5% significance level.

```python
import numpy as np
from scipy import stats
```

```python
# Given values
x1, s1, n1 = 1000, 100, 30
x2, s2, n2 = 950, 120, 30
```

```python
# Calculate t-statistic
numerator = x1 - x2
denominator = np.sqrt((s1*2 / n1) + (s2*2 / n2))
t_stat = numerator / denominator
```

```python
# Degrees of Freedom (Welch-Satterthwaite Equation)
df_numerator = (s1*2 / n1 + s2 / n2)*2
df_denominator = ((s1*2 / n1*2) / (n1 - 1) + ((s2 / n2)*2) / (n2 - 1))
df = df_numerator / df_denominator
```

```python
# p-value for two-tailed test
p_value = 2 * (1 - stats.t.cdf(abs(t_stat), df))
```

```python
# Critical t-value for 5% significance (two-tailed)
t_critical = stats.t.ppf(1 - 0.025, df)
```

```python
print(f"t-statistic: {t_stat:.2f}")
print(f"Degrees of Freedom: {df:.2f}")
print(f"p-value: {p_value:.5f}")
print(f"Critical t-value: ±{t_critical:.2f}")
```

```python
# Conclusion
if abs(t_stat) > t_critical:
    print("Reject the null hypothesis: Significant difference in average daily sales.")
else:
    print("Fail to reject the null hypothesis: No significant difference in average daily sales.")
```
